```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': ['H'],
    'G': [],
    'H': []
}

def dfs(graph, start):

    visited = []
    stack = []

    stack.append(start)

    while stack:

        node = stack.pop()

        if node not in visited:
            visited.append(node)
            print(node, end=" ")

            for neighbour in reversed(graph[node]):
                if neighbour not in visited:
                    stack.append(neighbour)


start = input("Enter Start Node: ").upper()

print("DFS Traversal:")
dfs(graph, start)
```
