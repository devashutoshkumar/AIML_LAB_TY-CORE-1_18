```python
import math
import numpy as np
import scipy
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Math Library

print("Square of 16: ", math.sqrt(16))
print("Cube of 8: ", math.cbrt(8))
print("Fact of 6:", math.factorial(5))
print("Power of 2:", math.pow(2,3))
print("Value of Pi:", math.pi)
```

```python
# NumPy Library

arr = np.array([10,20,30,40,50])

print("Array", arr)
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))
print("Min:", np.min(arr))
print("Std:", np.std(arr))
```

```python
# SciPy Library
from scipy import stats

data = [10,20,30,40,50]

print("Mean:", stats.tmean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=False).mode)
print("Variance:", np.var(data))
```

```python
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Rohan"],
    "Maths": [85, 90, 70, 95, 80],
    "Science": [80, 85, 75, 90, 88],
    "English": [78, 92, 72, 88, 85]
}

df = pd.DataFrame(data)

display(df)

print(df.head())

print(df.describe())

df['Total'] = df['Maths'] + df['Science'] + df['English']

df['Avg'] = df['Total']/3

display(df)
```

```python
plt.bar(df["Name"], df["Maths"])

plt.title("Maths Marks")
plt.xlabel("Student name")
plt.ylabel("marks")

plt.show()
```

```python
plt.plot(df["Name"], df["Science"], marker="o")

plt.title("Sc Marks")
plt.xlabel("Student name")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
marks = df[["Maths", "Science", "English"]]

sns.heatmap(marks.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation b/w subjects")
plt.show()
```

```python
sns.boxplot(data=marks)

plt.title("Distribution of student marks")
plt.ylabel("Marks")

plt.show()
```
